20) Diccionarios.

Tenemos un diccionario que contiene 10 tickers con su valor.

Tenemos un segundo diccionario que contiene los mismos 10 tickers con la cantidad de acciones que tenemos de ese ticker.

Usando un bucle, calcula cuanto vale mi cartera tienendo en cuenta que el valor es la suma de la cantidad de acciones de cada ticker * su valor

```python
# Ejericico largo
valores = {
    'AAPL': 174.55,
    'GOOGL': 138.20,
    'AMZN': 125.65,
    'TSLA': 265.75,
    'MSFT': 313.45,
    'NFLX': 405.50,
    'NVDA': 450.10,
    'META': 290.25,
    'INTC': 35.40,
    'IBM': 135.80
}

acciones = {
    'AAPL': 50,
    'GOOGL': 30,
    'AMZN': 20,
    'TSLA': 15,
    'MSFT': 40,
    'NFLX': 25,
    'NVDA': 10,
    'META': 35,
    'INTC': 60,
    'IBM': 45
}

# Escribe aquí tu código
cartera = 0
for ticker in valores:
  precio = valores[ticker]
  cantidad = acciones[ticker]

  cartera += precio * cantidad
print(cartera)
```

24) Gráficos de histograma y distribución de retornos

Descargar datos históricos de un activo, calcular los retornos y graficar el histograma de los retornos.

- Descarga datos históricos de un activo, elígelo tú.
- Calcula los retornos logarítmicos diarios.
- Calcula el histograma de los retornos logarítmicos.
- Superpone una distribución normal sobre el histograma.

```python
# Escribe aquí tu código
ticker = ["AMZN"]
data = yf.download(ticker, start = "2020-01-01", end = "2023-12-31", multi_level_index=False, auto_adjust=False)[["Adj Close"]]
msno.matrix(data)
```

```python
retornos = np.log(data).diff().dropna()
```

```python
from scipy.stats import norm

plt.figure(figsize=( 12,4))
plt.hist(retornos, bins = 100,edgecolor ="black", density = True)
x = np.linspace(retornos.min(), retornos.max(),100)
p = norm.pdf(x,retornos.mean(),retornos.std())
plt.plot(x,p)
plt.show()
```

```python
print(retornos.skew())
print(retornos.kurtosis())
```
